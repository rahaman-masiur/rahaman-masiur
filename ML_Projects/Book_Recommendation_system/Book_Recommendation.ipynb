```python
import numpy as np
import pandas as pd 
from sklearn.feature_extraction.text import CountVectorizer
```

```python
Books=pd.read_csv("data.csv")
print(Books.head())
print(Books.info())
```

```python
#Data preprocessing 
# Remove duplicates'
print(Books.duplicated().sum())
Books=Books.drop_duplicates(subset='title')
# Select important columns
#isbn13
#title	
#authors	
#categories	
#description
#average_rating

Books=Books[['isbn13','title','authors','categories','description']]

# Processsing null values
print(Books.isna().sum())
#remove the null values as its counts are not bigger
Books=Books.dropna()
Books.count()
Books.head()
Books = Books.reset_index()
```

```python
# Remove spaces between author name and categories name
Books['authors']=Books['authors'].str.replace(" ",'')

Books['authors']
```

```python
Books['categories']=Books['categories'].str.replace(" ",'')
Books['categories']
```

```python
Books['tags']=Books['categories']+"."+Books['description']+Books['authors']
Books['tags'][10]
Books=Books.drop(['categories','description','authors'],axis=1)
Books
```

```python
from nltk.stem.porter import PorterStemmer
stemmer=PorterStemmer()

def stem(text):
    y=[]
    for i in text.split():
        y.append(stemmer.stem(i))

    result=" ".join(y)
    return result
        
```

```python
# Apply stem function to our tag column
Books['tags']=Books['tags'].apply(stem)
Books['tags']
```

```python
cv =CountVectorizer(max_features=1000,stop_words='english')
vectors=cv.fit_transform(Books['tags']).toarray()
vectors
```

```python
cv.get_feature_names_out()
```

```python
vectors[0]
vectors.shape
```

```python
# Cosine distance
from sklearn.metrics.pairwise import cosine_similarity
distance_matrix=cosine_similarity(vectors)
print(distance_matrix)
```

```python
distance_matrix[1]
```

```python
Book_name=input("Enter name of a book")


idx=Books[Books['title']==Book_name].index[0]
print(idx)
distance=distance_matrix[idx]
print(distance)
```

```python

rec_books=sorted(list(enumerate(distance)),key=lambda x:x[1],reverse=True)[0:6]



def recommended(lst):
    for i in lst:
        print(Books.iloc[i[0]]['title'])

recommended(rec_books)
```
